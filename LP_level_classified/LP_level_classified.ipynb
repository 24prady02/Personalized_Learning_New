# LP level classification: how each question gives its level, and how the overall level is calculated

Twenty student answers about one small loop program (LP01 to LP20). **Part 1** shows, for every question and answer,
which rule gives its level. **Part 2** shows how one overall level is calculated for the whole answer. Each run has
its own folder under `runs/`, whose README works through the calculation with that run's numbers.

```java
int i, n = 5;
for (i = 1; i < n; i++)
    i++;
    System.out.println(i);
System.out.println(i);
```

(LP17 and LP18 use the program as the student first wrote it, with two missing `;` and lowercase `system`.)


# Part 1. How each question gives its level


## The rule for the level

A level is never asked from the model. The graph lists, for every concept, the **ways** in which an answer can show
each level (the indicators). The model only says which ways the student's words show, and which wrong rules the
student holds. The level then follows from the first of these rules that fits:

| Rule | When it applies | Level it gives | Source |
|---|---|---|---|
| (a) | A wrong rule (misconception) is held on the concept | **L0**, whatever else is shown | Lister et al. 2006, p. 2: a prestructural response manifests “a significant misconception of programming”; Teague & Lister 2014, p. 1; ProgMiscon (Chiodini et al. 2021), p. 2 |
| (b) | Only L0 ways are shown (everyday reading, inconsistent trace) | **L0** | Pea 1986, p. 2; Teague & Lister 2014, p. 1 (“cannot trace code”) |
| (c) | Otherwise | the **highest** level at which a way is shown; one shown way is enough | Lister et al. 2006, p. 2 (the SOLO levels build on each other); Jin, Lima & Wang 2025, p. 6 (an answer is scored at the level whose indicator it shows) |
| (d) | No way is shown and no mistake is held | **no level** | Workbook boundary: “a bare correct output is not levelled” |

The ways, with what the answer does and where each level stops:

| Level | Way (indicator) | What the answer does | Where the level stops |
|---|---|---|---|
| L0 | everyday reading | reads the code with the everyday or maths meaning of its words | any correct, programming-specific statement makes it L1 or higher |
| L0 | misconception | applies a documented wrong rule that the language disproves | a single slip in an otherwise rule-correct trace is L2 with an error |
| L0 | inconsistent trace | applies the right rule at one place and a wrong rule at another in the same trace | a slip under one consistent, correct rule is L2 with an error |
| L1 | names the element | names or describes the construct correctly | covering every part of the code in order is L2 |
| L1 | states the rule | states the rule, without applying it to the values | applying it to this code's values is L2 |
| L2 | line by line trace | executes each statement in order with its values | stating the purpose of the whole is L3 |
| L2 | output prediction | gives the output and justifies it value by value | a bare correct output is not levelled |
| L2 | one step execution | gives the change one execution makes | one step of a longer fragment is L1 |
| L3 | relation to neighbour | explains how the concept works with another construct in the same code | two separate correct rules are L1; running them in sequence is L2 |
| L3 | purpose of the whole | says what the code does for all inputs, deduced from the code | a purpose guessed from traced cases is L2 |
| L4 | predict a change | predicts what a change keeps or breaks, without re-tracing | re-tracing the changed code is L2 |
| L4 | diagnose an error | names the rule broken, the kind of error and the inputs that trigger it | fixing one given failing case is L3 |
| L4 | transfer to a new program | uses the construct in an unseen problem, knowing when it is the right tool | reproducing a familiar program is L1 to L3 |

Scoring by indicators, one concept at a time, follows Jin, Lima & Wang 2025, *EM:IP*, p. 2: “Analytic rubrics specify
indicators to be detected from students’ responses”, and p. 5: “analytical rubrics that contain item-specific
indicators for each LP level”. The levels and their ways are fixed before any answer is read (Jin et al. 2019, p. 6).

- **What the run gave for the whole answer.** no level (expected no level; match: yes).


# Part 2. How the overall level is calculated

One answer touches several concepts, and Part 1 gives each of them its own level. The overall level of the answer is
then worked out in four steps.

| Step | What is done | Source |
|---|---|---|
| 1. Take the components | The concepts the answer shows something about, each with its level from Part 1. Generic concepts (such as `statement`) are left out, unless a real mistake is held on them. | Jin et al. 2019, *IJSE*, p. 1325: “some researchers identify multiple progress variables of an LP, with each progress variable elaborating the development in one dimension of the LP”. Jin, Lima & Wang 2025, p. 5: multi-part answers “were combined into a single response to allow for a comprehensive analysis of student reasoning”. |
| 2. Is a real mistake held? | If the student holds a wrong rule or breaks a rule (the run names a root cause), the overall level is **L0**, however well the other components are done. | **This project's own step; no paper states it.** It applies rule (a) of Part 1 (Lister et al. 2006, p. 2) to the whole answer. |
| 3. If no mistake, count | Count the components at each level. The level with the **largest share** is the overall level; on a tie, the lower level. No component, no level. | Jin et al. 2019, pp. 1338–1339, Table 4: level shares per student, for example Student 210: L2 .68, L3 .32; Student 220: L1 .50, L2 .50, described as “rely on Level 1 and Level 2 reasoning”. The tie rule is an adaptation: one number is needed, so the lower is taken. |
| 4. Check coherence | Highest component level minus lowest. At most 1: **coherent**. 2 or more: **LOW coherence**, the one number hides real differences. | Jin et al. 2019, p. 1338: “each student produced responses at adjacent levels, suggesting that those students’ ideas are coherent to a certain degree”; p. 1337: “If the result shows evidence of ‘knowledge-in-pieces’ … we cannot make any inferences about the students’ levels”. |

Jin et al. apply shares across many items and students with a partial credit model (Jin et al. 2017, *JRST*,
pp. 862–863); here the same idea is applied across the components of one answer with simple shares.


## Two worked examples

**LP15, no mistake held.**
1. Components: `for_loop` L3, `increment_operators` L3, `loop_body` L2, `loop_control` L3, `loop_variable` L3. Five components.
2. No wrong rule is held.
3. Count: 1 of 5 at L2 (1 ÷ 5 = 0.20), 4 of 5 at L3 (4 ÷ 5 = 0.80). The largest share is at L3.
4. Highest L3, lowest L2: 3 − 2 = 1, coherent.

Overall level: **L3**.

**LP04, a mistake held.**
1. Components: `comparison_operators` L2, `control_flow` L2, `for_loop` L2, `increment_operators` L0, `loop_body` L0, `loop_control` L2, `loop_variable` L0. Seven components.
2. The student holds the wrong rule BodyChangeIgnoredByCondition at `loop_variable`. So the overall level is L0.
3. Count, for information: 3 of 7 at L0 (0.43), 4 of 7 at L2 (0.57). The largest share would be L2, but step 2 decides.
4. Highest L2, lowest L0: 2 − 0 = 2, LOW coherence: the student traces most parts correctly and holds one wrong rule.

Overall level: **L0**.
